## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("/content/content_refresh_anonymized.csv")

df.head()

total_rows = len(df)
total_clients = df["client_id"].nunique()
# 1. Archetype 1: Hidden Gems (Page 1 rank, high impressions, low CTR)
hidden_gems = df[
    (df["avg_position"] > 0)
    & (df["avg_position"] <= 10)
    & (df["impressions_90d"] >= 500)
    & (df["ctr"] < 1.0)
]
# 2. Archetype 2: High-Demand Decaying Assets
declining_demand = df[
    (df["impressions_90d"] >= 500) & (df["trend_direction"] == "down")
]
impressions_at_risk = declining_demand["impressions_90d"].sum()
# 3. Archetype 3: Dead-Weight / Low Demand Tail
low_demand = df[df["impressions_90d"] < 10]
print("=== STARTER DATA EVIDENCE ===")
print(
    f"Total Content Items (Grain = 1 page): {total_rows:,} across"
    f" {total_clients} clients\n"
)
print(
    f"1. Hidden Gems (Page 1 rank, >=500 imp, CTR < 1.0%): {len(hidden_gems):,}"
    f" pages ({len(hidden_gems)/total_rows*100:.1f}%)"
)
print(
    "2. High-Demand Assets in Active Decline (>=500 imp, trend 'down'):"
    f" {len(declining_demand):,} pages"
)
print(
    f"   -> Total 90-day impressions at risk in this group:"
    f" {impressions_at_risk:,}"
)
print(
    f"3. Low/Zero Demand Tail (<10 imp in 90 days): {len(low_demand):,} pages"
    f" ({len(low_demand)/total_rows*100:.1f}%)"
)
```
